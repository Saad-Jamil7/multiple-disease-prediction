```python
"""
Train a Breast Cancer prediction model.

Dataset: Wisconsin Diagnostic Breast Cancer (WDBC) dataset, bundled directly
with scikit-learn (sklearn.datasets.load_breast_cancer) - 569 samples,
30 numeric features computed from digitized images of a breast mass
fine needle aspirate (FNA).

Target (as provided by sklearn):
    0 = malignant
    1 = benign
We keep sklearn's original encoding so it's easy to cross-check.

We also write out breast_cancer.csv to the project folder, mirroring how
diabetes.csv / heart.csv / parkinsons.csv are checked into the repo.
"""
```

```python
import pickle
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
# ---------------------------------------------------------------------------
# Load data
# ---------------------------------------------------------------------------
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
Y = pd.Series(data.target, name='target')
```

```python
# save a csv copy alongside the other datasets in the repo
csv_df = X.copy()
csv_df['target'] = Y
csv_df.to_csv('breast_cancer.csv', index=False)
```

```python
# ---------------------------------------------------------------------------
# Train / test split
# ---------------------------------------------------------------------------
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, stratify=Y, random_state=2
)
```

```python
# ---------------------------------------------------------------------------
# Train model (linear SVM, consistent with the rest of the project)
# ---------------------------------------------------------------------------
classifier = svm.SVC(kernel='linear')
classifier.fit(X_train, Y_train)
```

```python
train_acc = accuracy_score(Y_train, classifier.predict(X_train))
test_acc = accuracy_score(Y_test, classifier.predict(X_test))
print('Breast Cancer Model')
print('Accuracy score of the training data :', train_acc)
print('Accuracy score of the test data     :', test_acc)
print('Feature order (must match Streamlit form order):')
print(list(X.columns))
```

```python
# ---------------------------------------------------------------------------
# Save model
# ---------------------------------------------------------------------------
filename = 'breast_cancer_model.sav'
pickle.dump(classifier, open(filename, 'wb'))
print(f"Saved model to {filename}")
```

```python
# quick sanity check reload
loaded_model = pickle.load(open(filename, 'rb'))
sample = np.asarray(X_test.iloc[0]).reshape(1, -1)
print('Sample prediction:', loaded_model.predict(sample), 'true label:', Y_test.iloc[0])
```
